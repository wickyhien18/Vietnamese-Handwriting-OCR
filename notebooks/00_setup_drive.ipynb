# 00 - Set up the shared Drive folder

Run this notebook **once by the person who owns the Drive folder**. It creates the `ml-project` folder structure in My Drive.

Afterwards that person shares the `ml-project` folder with the other two as **Editor** (Drive web page: right click, Share).
Everyone else runs only cell 3 to check that they can read and write the folder.


## Cell 1: mount Drive and create folders

```
from google.colab import drive
import os

drive.mount("/content/drive")

DRIVE_ROOT = "/content/drive/MyDrive/ml-project"
SUBDIRS = [
    "data/uit_hwdb",
    "data/corpus",
    "checkpoints/ocr",
    "checkpoints/corrector",
    "outputs",
]

for sub in SUBDIRS:
    os.makedirs(os.path.join(DRIVE_ROOT, sub), exist_ok=True)

for base, dirs, files in os.walk(DRIVE_ROOT):
    depth = base[len(DRIVE_ROOT):].count(os.sep)
    print("  " * depth + os.path.basename(base) + "/")
```

## Cell 2: (teammates only) if the folder is shared with you

A folder shared with you appears under *Shared with me*, not under *My Drive*. In the Drive web page right click `ml-project`, choose *Organize* then *Add shortcut*, and put the shortcut in *My Drive*. Then `DRIVE_ROOT` above works for you too.
If the path still does not exist, find the real path in the file browser on the left and change `DRIVE_ROOT`.


## Cell 3: check read and write access

```
test_path = os.path.join(DRIVE_ROOT, "outputs", ".write_test")
with open(test_path, "w") as handle:
    handle.write("ok")
os.remove(test_path)
print("Drive folder is readable and writable.")
```
